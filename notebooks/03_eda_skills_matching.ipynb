# 🧭 O*NET Skills Taxonomy & Career Pathway Matching

Exploring occupational standards, essential skills, and software requirements to build skill gap analysis.

```python
import pandas as pd
import numpy as np
import sys
from pathlib import Path

BASE_DIR = Path('..').resolve()
sys.path.append(str(BASE_DIR))
from app.backend.config import *
```

### 1. Load Processed Skills Taxonomy

```python
df_skills = pd.read_csv(DATA_PROCESSED_DIR / 'processed_skills_taxonomy.csv')
print(f'Total Occupations in Taxonomy: {len(df_skills)}')
df_skills.head(10)[['O*NET-SOC Code', 'Title', 'Description']]
```

### 2. Search Occupations and Inspect Essential Competencies

```python
def inspect_role(search_query):
    match = df_skills[df_skills['Title'].str.contains(search_query, case=False, na=False)]
    if match.empty:
        print(f'No roles found for query: {search_query}')
        return
    for _, row in match.head(3).iterrows():
        print('=' * 60)
        print(f"Title: {row['Title']} ({row['O*NET-SOC Code']})")
        print(f"Description: {row['Description'][:150]}...")
        print(f"Top Skills: {str(row['TopSkills'])[:100]}...")
        print(f"Software Tools: {str(row['SoftwareTools'])[:100]}...")

inspect_role('Software')
inspect_role('Manager')
```
